```python
import ultralytics
from ultralytics import YOLO

model = YOLO('yolo11n-obb.pt') # load pretrained model

print(type(model.names), len(model.names))
print(model.names)
```

```python
model.train(data='/content/dataset/data.yaml', epochs=500, patience=5, imgsz=640)
```

```python
print(type(model.names), len(model.names))
print(model.names)
```

```python
results=model(source='/content/dataset/test/images', save=True)
```
